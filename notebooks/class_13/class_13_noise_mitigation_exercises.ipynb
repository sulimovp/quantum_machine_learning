```python3
# YOUR CODE HERE: Build confusion matrix and apply mitigation
print("Measurement error mitigation for 1 qubit.")
print("Setup: Readout error with probability p=0.05")

p_readout = 0.05

# YOUR CODE HERE: Construct confusion matrix M
M_calib = None

if M_calib is not None:
    print(f"\nConfusion matrix M:")
    print(M_calib)
    
    M_inv = np.linalg.inv(M_calib)
    print(f"\nInverse M^-1:")
    print(np.around(M_inv, 6))
else:
    print("...complete the matrix construction...")
```

```python3
# YOUR CODE HERE: Simulate measurement error on Hadamard circuit
num_shots = 10000
ideal_counts_h = np.array([num_shots // 2, num_shots // 2])

# YOUR CODE HERE: Simulate corrupted counts with readout error
measured_counts = None
measured_probs = None

if measured_counts is not None:
    print(f"Ideal counts for H: {ideal_counts_h}")
    print(f"Measured counts (with error): {measured_counts}")
    print(f"Measured probabilities: {measured_probs}")
else:
    print("...complete the error simulation...")
```

```python3
# YOUR CODE HERE: Apply mitigation
if measured_counts is not None and M_inv is not None:
    mitigated_counts = M_inv @ measured_counts
    mitigated_probs = mitigated_counts / np.sum(mitigated_counts)
    
    print(f"Mitigated counts: {np.around(mitigated_counts, 1)}")
    print(f"Mitigated probabilities: {np.around(mitigated_probs, 6)}")
    
    ideal_probs = np.array([0.5, 0.5])
    
    error_before = np.linalg.norm(measured_probs - ideal_probs)
    error_after = np.linalg.norm(mitigated_probs - ideal_probs)
    
    print(f"\nError before: {error_before:.6f}")
    print(f"Error after: {error_after:.6f}")
else:
    print("...complete earlier steps first...")
```

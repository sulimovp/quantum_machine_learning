## 1. QML Landscape: Five Paradigms

Quantum machine learning spans five distinct paradigms, each with different strengths, hardware requirements, and use cases:

### Paradigm 1: Quantum-Inspired Algorithms
Classical algorithms inspired by quantum concepts (entanglement, superposition). No quantum hardware required. Examples: tensor networks, variational classical circuits.

### Paradigm 2: Quantum-Enhanced Classical Learning
Hybrid systems that delegate specific tasks to quantum processors (e.g., kernel computation, sampling). Short quantum circuits, near-term device compatible. Examples: quantum kernel SVM, quantum feature maps.

### Paradigm 3: Quantum-Native Algorithms
Fully quantum algorithms that exploit quantum properties throughout. Examples: VQE, QAOA, Quantum Born Machines. Moderate circuit depth, NISQ-era compatible.

### Paradigm 4: Quantum Physics-Informed Neural Networks (QPINNs)
Embed physical constraints (PDEs) into parameterized quantum circuits. Solve differential equations and scientific computing problems. Parameter-shift rule for derivatives.

### Paradigm 5: Quantum-Inspired Neural Networks (QINNs)
Classical neural networks with quantum-inspired architectures and initialization schemes. No quantum hardware. Fast training, good interpretability.

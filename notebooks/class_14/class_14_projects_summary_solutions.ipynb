## 1. QML Landscape: Five Paradigms

Quantum machine learning spans five distinct paradigms, each with different strengths, hardware requirements, and use cases:

### Paradigm 1: Quantum-Inspired Algorithms
Classical algorithms inspired by quantum concepts (entanglement, superposition). No quantum hardware required. Examples: tensor networks, variational classical circuits.

### Paradigm 2: Quantum-Enhanced Classical Learning
Hybrid systems that delegate specific tasks to quantum processors (e.g., kernel computation, sampling). Short quantum circuits, near-term device compatible. Examples: quantum kernel SVM, quantum feature maps.

### Paradigm 3: Quantum-Native Algorithms
Fully quantum algorithms that exploit quantum properties throughout. Examples: VQE, QAOA, Quantum Born Machines. Moderate circuit depth, NISQ-era compatible.

### Paradigm 4: Quantum Physics-Informed Neural Networks (QPINNs)
Embed physical constraints (PDEs) into parameterized quantum circuits. Solve differential equations and scientific computing problems. Parameter-shift rule for derivatives.

### Paradigm 5: Quantum-Inspired Neural Networks (QINNs)
Classical neural networks with quantum-inspired architectures and initialization schemes. No quantum hardware. Fast training, good interpretability.

## 2. Decision Framework for QML Approach Selection

Use this framework to decide which QML paradigm is most appropriate for your problem.

```python
def recommend_qml_approach(data_type, n_samples, n_features, task_type):
    """
    Recommend a QML approach based on problem characteristics.
    
    Args:
        data_type: 'tabular', 'image', 'text', 'graph', or 'scientific'
        n_samples: number of training samples
        n_features: feature dimension or problem size
        task_type: 'classification', 'regression', 'optimization', 'sampling', or 'pde'
    
    Returns:
        str: recommended approach and rationale
    """
    
    recommendation = "Recommended QML Approach:\n"
    
    # Decision tree based on problem characteristics
    if task_type == 'pde':
        return recommendation + "QPINN (Quantum Physics-Informed Neural Network)\nRationale: PDE solving requires embedding physical constraints. Use parameter-shift rule for derivatives."
    
    if task_type == 'optimization':
        if n_features <= 20:
            return recommendation + "QAOA (Quantum Approximate Optimization Algorithm)\nRationale: Combinatorial problems with small instance size benefit from QAOA's quantum speedup potential."
        else:
            return recommendation + "Classical Optimization (e.g., GradientDescent, GeneticAlgorithm)\nRationale: Problem size too large for current quantum devices; classical methods more practical."
    
    if task_type == 'sampling':
        return recommendation + "Quantum Born Machine\nRationale: Quantum circuits excel at sampling from complex distributions. Requires expressive entangling layers."
    
    if task_type == 'classification' or task_type == 'regression':
        if n_samples > 10000 or n_features > 50:
            return recommendation + "QINN (Quantum-Inspired Neural Network) as baseline; then try Quantum Kernel SVM for comparison\nRationale: Large datasets favor classical methods. Use QINN for quantum-inspired architecture."
        elif n_features > 20:
            return recommendation + "Quantum Kernel SVM (Quantum-Enhanced)\nRationale: Leverage quantum feature maps for moderate feature dimensions. Classical SVM readout."
        else:
            return recommendation + "Variational Quantum Classifier (VQC)\nRationale: Low-dimensional problems allow full quantum training. Use parameterized circuits with classical optimization."
    
    return recommendation + "Classical machine learning baseline\nRationale: Problem characteristics suggest hybrid or classical approaches are more suitable."

# Example calls
print("\nExample Recommendations:")
print("\n" + "-"*80)
print("\n1. Small tabular classification, 500 samples, 8 features:")
print(recommend_qml_approach('tabular', 500, 8, 'classification'))

print("\n" + "-"*80)
print("\n2. MaxCut optimization on a 15-node graph:")
print(recommend_qml_approach('graph', 1, 15, 'optimization'))

print("\n" + "-"*80)
print("\n3. Solve Schrodinger equation (PDE):")
print(recommend_qml_approach('scientific', 100, 1, 'pde'))

print("\n" + "-"*80)
print("\n4. Large tabular dataset, 50000 samples, 30 features:")
print(recommend_qml_approach('tabular', 50000, 30, 'classification'))

print("\n" + "-"*80)
print("\n5. Generate samples from a learned distribution:")
print(recommend_qml_approach('scientific', 5000, 16, 'sampling'))

print("\n" + "-"*80)
```
